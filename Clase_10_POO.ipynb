```python
class Casa:
    """
    Construir el constructor
    argumentos:
      color(Str): color de la casa
      numero_habitaciones(int): el número de habitaciones en la casa
      numero_baños(int): el número de baños de la casa
      estacionamiento(bool): si tiene estacionamiento o no
      tiene_patio(bool): si tiene patio o no
    """
    def __init__(self, color, numero_habitaciones, numero_ducha, estacionamiento, tiene_patio):
        self.color = color
        self.numero_habitaciones = numero_habitaciones
        self.numero_ducha = numero_ducha
        self.estacionamiento = estacionamiento
        self.tiene_patio = tiene_patio

    def descripcion(self):
        descripcion = f"la casa es de color {self.color} tiene {self.numero_habitaciones} habitaciones y tiene {self.numero_ducha} duchas"
        if self.estacionamiento:
            descripcion += " tiene estacionamiento"
        if self.tiene_patio: # Corrected variable name from patio to tiene_patio
            descripcion += " tiene patio"
        return descripcion

    def pintar(self, nuevo_color):
        self.color = nuevo_color
        print(f"la casa esta pintada de color {self.color}")
```

```python
mi_casa = Casa("Marron",3,4,True,False)
type(mi_casa)
```

```python
print(mi_casa.descripcion())
```

```python
mi_casa.tiene_patio=False
print(mi_casa.descripcion())
```
